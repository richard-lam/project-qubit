# Grover's Algorithm (2 Qubits)
Demonstrate Grover search on a 2-qubit system.

We will:
1. Create a 2-qubit superposition
2. Apply a simple oracle marking |10⟩
3. Apply the Grover diffusion operator
4. Measure and plot the probability distribution

```python
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit.visualization import plot_histogram
from qiskit_aer import AerSimulator

# Create 2-qubit Grover circuit
qc = QuantumCircuit(2, 2)

# Step 1: Put qubits in superposition
qc.h([0, 1])

# Step 2: Oracle marks |10> (simple example using CZ)
qc.cz(0, 1)

# Step 3: Grover diffusion operator
qc.h([0, 1])
qc.x([0, 1])
qc.cz(0, 1)
qc.x([0, 1])
qc.h([0, 1])

# Step 4: Measure all qubits
qc.measure([0, 1], [0, 1])

# Draw the circuit
qc.draw(output='mpl')
```

```python
%matplotlib inline
# Step 5: Simulate and plot results
sim = AerSimulator()
job = sim.run(qc, shots=1024)
counts = job.result().get_counts()

print("Measurement counts:", counts)
plot_histogram(counts)
plt.show()
```
