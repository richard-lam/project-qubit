# CNOT Gate
This notebook demonstrates a 2-qubit CNOT gate and how it creates entanglement.

We will:
1. Create a 2-qubit circuit and apply gates
2. Visualize the circuit
3. Inspect the statevector
4. Optionally visualize qubits on the Bloch sphere

## Create and visualize the 2-qubit circuit
We apply a Hadamard gate to qubit 0 and a CNOT gate to entangle qubit 0 and 1.

```python
from qiskit import QuantumCircuit
import matplotlib.pyplot as plt

# Create 2-qubit circuit and apply gates
qc = QuantumCircuit(2)
qc.h(0)      # Put qubit 0 in superposition
qc.cx(0, 1)  # CNOT entangles qubit 0 and 1

# Visualize the circuit using Matplotlib
qc.draw(output='mpl')
```

## Inspect the statevector
We can extract the quantum state after applying the gates without running a simulator.

```python
%matplotlib inline
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector

# Get the statevector from the circuit
state = Statevector.from_instruction(qc)

print("Statevector:", state)

# Optional: visualize Bloch vectors of each qubit
plot_bloch_multivector(state)
plt.show()
```
