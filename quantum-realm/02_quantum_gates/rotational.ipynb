# Single-Qubit Rotation Gates

This notebook demonstrates how rotation gates RX, RY, and RZ affect a single qubit.
We will:
1. Create a single qubit circuit
2. Apply RX, RY, RZ rotations
3. Visualize the circuit
4. Observe how the qubit state changes on the Bloch sphere

```python
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector

# Create a single qubit circuit
qc = QuantumCircuit(1)

# Apply rotations
qc.rx(3.14159/2, 0)  # 90 degrees around X-axis
qc.ry(3.14159/4, 0)  # 45 degrees around Y-axis
qc.rz(3.14159/2, 0)  # 90 degrees around Z-axis

# Draw the circuit diagram
qc.draw(output='mpl')
```

## Simulate and visualize the final qubit state

We use `Statevector.from_instruction(qc)` to directly get the qubit state before measurement.

```python
%matplotlib inline
# Get statevector directly from the circuit
state = Statevector.from_instruction(qc)

# Print statevector
print("Statevector:", state)

# Plot Bloch vector
plot_bloch_multivector(state)
plt.show()
```
