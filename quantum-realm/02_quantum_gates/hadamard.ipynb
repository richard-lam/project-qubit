# Hadamard Gate on a Single Qubit

This notebook demonstrates how the Hadamard gate puts a single qubit into superposition.
We will:
1. Create a single qubit circuit
2. Apply the Hadamard gate
3. Visualize the circuit
4. Observe the qubit on the Bloch sphere

```python
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_bloch_multivector

# Create a single qubit circuit and apply Hadamard
qc = QuantumCircuit(1)
qc.h(0)  # Apply Hadamard gate

# Draw the circuit diagram
qc.draw(output='mpl')
```

## Simulate the statevector and visualize on Bloch sphere

```python
%matplotlib inline
# Simulate state vector
state = Statevector.from_instruction(qc)

# Print the statevector
print("Statevector:", state)

# Plot Bloch vector
plot_bloch_multivector(state)
plt.show()
```
