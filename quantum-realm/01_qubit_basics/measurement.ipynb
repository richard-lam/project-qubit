# Qubit Measurement and Probabilities

This notebook demonstrates repeated measurements of a single qubit in superposition.
We will:
1. Create a qubit in superposition using a Hadamard gate
2. Measure the qubit multiple times
3. Plot the resulting probability distribution

```python
%matplotlib inline
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit.visualization import plot_histogram
from qiskit_aer import AerSimulator

# Step 1: Create a single-qubit circuit
qc = QuantumCircuit(1, 1)
qc.h(0)  # Put qubit in superposition
qc.measure(0, 0)  # Measure the qubit

# Draw the circuit
qc.draw(output='mpl')
plt.show()
```

```python
# Step 2: Simulate measurements
sim = AerSimulator()
job = sim.run(qc, shots=5000)
result = job.result()
counts = result.get_counts()

# Step 3: Show results
print("Measurement counts:", counts)
plot_histogram(counts)
plt.show()
```
