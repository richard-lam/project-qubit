# Superposition
This notebook demonstrates how a single qubit can be placed into superposition using the Hadamard gate.
We will:
1. Create a quantum circuit
2. Simulate measurements using the statevector

## Create a qubit in superposition

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_histogram

# Create a single-qubit circuit and apply Hadamard
qc = QuantumCircuit(1)
qc.h(0)
qc.draw(output='mpl')
```

## Simulate Measurement
We sample measurements directly from the quantum state. This mimics running the circuit many times.

```python
# Get statevector
state = Statevector.from_instruction(qc)

# Sample measurement results
shots = 1024
counts = state.sample_counts(shots=shots)

print(counts)

# Plot histogram of measurement results
plot_histogram(counts)
```
